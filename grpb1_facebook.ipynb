```python
df=pd.read_csv('dataset_facebook.csv',sep=';')
print("Original Data")
print(df.head())
```

```python
# A)Create data subsets

#first 5 rows
subset1=df.head(5)
subset2=df[['Type','Category','like']]
print(subset2.head())
```

```python
# B)Merge Data

#creating second dataframe
extra_data = pd.DataFrame({
    'Type': ['Photo', 'Video', 'Status'],
    'Content': ['Image Post', 'Video Post', 'Text Post']
})
# Merge both dataframes using common column 'Type'
merged_data = pd.merge(df, extra_data, on='Type')

print("\nMerged Data")
print(merged_data.head())
```

```python
# =====================================================
# c) Sort Data
# =====================================================

# Sort dataset according to likes
sorted_data = df.sort_values(by='like', ascending=False)

print("\nSorted Data")
print(sorted_data[['Type', 'like']].head())
```

```python
# =====================================================
# d) Transposing Data
# =====================================================

# Convert rows into columns
transpose_data = subset1.T

print("\nTransposed Data")
print(transpose_data)
```

```python
# =====================================================
# e) Shape and Reshape Data
# =====================================================

# Shape gives number of rows and columns
print("\nShape of Dataset")
print(df.shape)


# Reshape using melt()
# Converts columns into row format
reshape_data = pd.melt(
    df,
    id_vars=['Type'],
    value_vars=['like', 'share']
)

print("\nReshaped Data")
print(reshape_data.head())
```
