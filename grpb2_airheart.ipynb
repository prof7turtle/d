```python
# =====================================================
# Import Required Libraries
# =====================================================

import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression


# =====================================================
# Load Datasets
# =====================================================

# Read Air Quality dataset
air = pd.read_csv("air_quality.csv")

# Read Heart Disease dataset
heart = pd.read_csv("heart.csv")

print("Air Quality Dataset")
print(air.head())

print("\nHeart Disease Dataset")
print(heart.head())


# =====================================================
# a) Data Cleaning
# =====================================================

# Remove rows having missing values
# .copy() avoids warning messages
air_clean = air.dropna().copy()

heart_clean = heart.dropna().copy()

print("\nCleaned Air Dataset")
print(air_clean.head())


# =====================================================
# b) Data Integration
# =====================================================

# Create common ID column in both datasets
air_clean.loc[:, 'ID'] = range(1, len(air_clean) + 1)

heart_clean.loc[:, 'ID'] = range(1, len(heart_clean) + 1)

# Merge both datasets using ID column
integrated_data = pd.merge(air_clean, heart_clean, on='ID')

print("\nIntegrated Data")
print(integrated_data.head())


# =====================================================
# c) Data Transformation
# =====================================================

# Create MinMaxScaler object
scaler = MinMaxScaler()

# Select numerical columns only
numeric_cols = integrated_data.select_dtypes(include='number').columns

# Transform data between 0 and 1
integrated_data[numeric_cols] = scaler.fit_transform(
    integrated_data[numeric_cols]
)

print("\nTransformed Data")
print(integrated_data.head())


# =====================================================
# d) Error Correcting
# =====================================================

# Remove duplicate rows
corrected_data = integrated_data.drop_duplicates()

print("\nCorrected Data")
print(corrected_data.head())


# =====================================================
# e) Data Model Building
# =====================================================

# Input data (all columns except target)
X = heart_clean.drop('target', axis=1)

# Output column
y = heart_clean['target']

# Standardize data for better training
std_scaler = StandardScaler()

X_scaled = std_scaler.fit_transform(X)

# Create Logistic Regression model
model = LogisticRegression(max_iter=5000)

# Train the model
model.fit(X_scaled, y)

print("\nModel Built Successfully")
```
