```python
# =====================================================
# VISUALIZATION OF FOREST FIRE DATASET
# USING MATPLOTLIB & SEABORN
# =====================================================

# Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# =====================================================
# LOAD DATASET
# =====================================================

forest_df = pd.read_csv("forestfires.csv")

print("First 5 Rows:")
print(forest_df.head())

# =====================================================
# 1. HISTOGRAM
# =====================================================

plt.figure(figsize=(8,5))

plt.hist(forest_df['temp'], bins=20)

plt.title("Temperature Distribution")
plt.xlabel("Temperature")
plt.ylabel("Frequency")

plt.show()

# =====================================================
# 2. BAR PLOT
# =====================================================

plt.figure(figsize=(10,5))

sns.barplot(
    x='month',
    y='area',
    data=forest_df
)

plt.title("Monthwise Burned Area")
plt.xlabel("Month")
plt.ylabel("Burned Area")

plt.show()

# =====================================================
# 3. SCATTER PLOT
# =====================================================

plt.figure(figsize=(8,5))

plt.scatter(
    forest_df['temp'],
    forest_df['area']
)

plt.title("Temperature vs Burned Area")
plt.xlabel("Temperature")
plt.ylabel("Burned Area")

plt.show()

# =====================================================
# 4. LINE PLOT
# =====================================================

plt.figure(figsize=(10,5))

plt.plot(forest_df['temp'])

plt.title("Temperature Variation")
plt.xlabel("Index")
plt.ylabel("Temperature")

plt.show()

# =====================================================
# 5. BOX PLOT
# =====================================================

plt.figure(figsize=(6,5))

sns.boxplot(y=forest_df['wind'])

plt.title("Wind Speed Distribution")

plt.show()

# =====================================================
# 6. HEATMAP
# =====================================================

plt.figure(figsize=(10,8))

sns.heatmap(
    forest_df.corr(numeric_only=True),
    annot=True
)

plt.title("Correlation Heatmap")

plt.show()
```
